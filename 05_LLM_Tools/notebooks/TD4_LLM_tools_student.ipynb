## 0. Setup

```python
import numpy as np
import matplotlib.pyplot as plt
RANDOM_STATE=42
rng=np.random.default_rng(RANDOM_STATE)

# A synthetic 1D signal with known peaks (positions we will check against)
x=np.linspace(0,100,1000)
true_centers=[15,32,48,70,86]
signal=np.zeros_like(x)
for c in true_centers:
    signal+=np.exp(-0.5*((x-c)/1.2)**2)*rng.uniform(0.6,1.0)
signal+=rng.normal(0,0.03,x.size)  # noise
plt.plot(x,signal,color="#028090"); plt.title("Signal with 5 peaks"); plt.xlabel("x"); plt.ylabel("intensity"); plt.show()
print("true peak centres (x):", true_centers)
```
